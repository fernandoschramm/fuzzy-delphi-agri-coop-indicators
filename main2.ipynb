```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Parameters
# ------------------------------------------------------------
INPUT_FILE = "dataset.xlsx"
OUTPUT_FILE = "results.xlsx"
FIG_DIR = Path("Fig")

RELEVANCE_THRESHOLD = 7.0
CV_THRESHOLD = 10.0

FIG_DIR.mkdir(exist_ok=True)

# ------------------------------------------------------------
# Load dataset
# ------------------------------------------------------------
df = pd.read_excel(INPUT_FILE, sheet_name="Evaluations")

# Keep only identification columns and expert-evaluation columns
id_cols = ["ID", "Dimension", "Indicator"]

expert_cols = {
    1: ("Expert 1 (aij)", "Expert 1 (bij)", "Expert 1 (cij)"),
    2: ("Expert 2 (aij)", "Expert 2 (bij)", "Expert 2 (cij)"),
    3: ("Expert 3 (aij)", "Expert 3 (bij)", "Expert 3 (cij)"),
    4: ("Expert 4 (aij)", "Expert 4 (bij)", "Expert 4 (cij)"),
    5: ("Expert 5 (aij)", "Expert 5 (bij)", "Expert 5 (cij)")
}

evaluation_cols = [col for cols in expert_cols.values() for col in cols]
df = df[id_cols + evaluation_cols].copy()

# ------------------------------------------------------------
# Data validation
# ------------------------------------------------------------
# Convert expert evaluations to numeric values
for col in evaluation_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Check missing values
if df[evaluation_cols].isna().any().any():
    missing = df[df[evaluation_cols].isna().any(axis=1)][id_cols]
    raise ValueError(f"Missing or non-numeric values found in: \n{missing}")

# Check 0-10 scale
if ((df[evaluation_cols] < 0) | (df[evaluation_cols] > 10)).any().any():
    raise ValueError("Some expert evaluations are outside the 0-10 scale.")

# Check triangular fuzzy-number consistency: aij <= bij <= cij
for expert, (a_col, b_col, c_col) in expert_cols.items():
    invalid = df[(df[a_col] > df[b_col]) | (df[b_col] > df[c_col])]
    if not invalid.empty:
        raise ValueError(
            f"Inconsistent triangular fuzzy numbers for Expert {expert}. "
            f"Check aij <= bij <= cij for indicators: {invalid['ID'].tolist()}"
        )

# ------------------------------------------------------------
# Individual defuzzification
# ------------------------------------------------------------
def defuzzify_individual(a, b, c):
    return (a + 4*b + c) / 6

for expert, (a_col, b_col, c_col) in expert_cols.items():
    df[f"A{expert}j"] = defuzzify_individual(
        df[a_col],
        df[b_col],
        df[c_col]
    )

A_cols = [f"A{expert}j" for expert in expert_cols.keys()]

# ------------------------------------------------------------
# Relevance and consensus measures
# ------------------------------------------------------------
df["mu_j"] = df[A_cols].mean(axis=1)
df["median_j"] = df[A_cols].median(axis=1)
df["sigma_j"] = df[A_cols].std(axis=1, ddof=0)
df["CV_j"] = df["sigma_j"] / df["mu_j"]
df["CV_j (%)"] = 100 * df["CV_j"]

# Combined selection criterion
df["Selected"] = (df["mu_j"] >= RELEVANCE_THRESHOLD) & (df["CV_j (%)"] <= CV_THRESHOLD)

# Optional classification for interpretation
df["Selection status"] = np.where(df["Selected"], "Selected", "Not selected")

# ------------------------------------------------------------
# Summary tables
# ------------------------------------------------------------
summary_by_dimension = (
    df.groupby("Dimension")
      .agg(
          Initial_indicators=("ID", "count"),
          Selected_indicators=("Selected", "sum"),
          Mean_relevance=("mu_j", "mean"),
          Median_relevance=("median_j", "median"),
          Mean_CV_percent=("CV_j (%)", "mean")
      )
      .reset_index()
)

summary_by_dimension["Retention_rate (%)"] = (
    100 * summary_by_dimension["Selected_indicators"] / summary_by_dimension["Initial_indicators"]
)

# Sensitivity analysis for alternative relevance thresholds
thresholds = [7.0, 7.25, 7.5, 7.75, 8.0]
sensitivity = []

for threshold in thresholds:
    selected_temp = (df["mu_j"] >= threshold) & (df["CV_j (%)"] <= CV_THRESHOLD)
    sensitivity.append({
        "Relevance threshold": threshold,
        "CV threshold (%)": CV_THRESHOLD,
        "Selected indicators": int(selected_temp.sum())
    })

sensitivity_df = pd.DataFrame(sensitivity)

# Selected indicators only
selected_df = df[df["Selected"]].copy()

# ------------------------------------------------------------
# Export results.xlsx
# ------------------------------------------------------------
with pd.ExcelWriter(OUTPUT_FILE, engine="openpyxl") as writer:
    df.to_excel(writer, sheet_name="All_results", index=False)
    selected_df.to_excel(writer, sheet_name="Selected_indicators", index=False)
    summary_by_dimension.to_excel(writer, sheet_name="Summary_by_dimension", index=False)
    sensitivity_df.to_excel(writer, sheet_name="Sensitivity", index=False)

print(f"Total indicators: {len(df)}")
print(f"Selected indicators: {df['Selected'].sum()}")
print(summary_by_dimension)
print(f"Results saved to: {OUTPUT_FILE}")
```

```python
# ------------------------------------------------------------
# Figure 1: CV dispersion plot
# ------------------------------------------------------------
plt.figure(figsize=(18, 7))
plt.scatter(df["ID"], df["CV_j (%)"])
plt.axhline(y=CV_THRESHOLD, linestyle="--", label=f"{CV_THRESHOLD:.0f}% CV threshold")
plt.xticks(rotation=90)
plt.xlabel("Indicators")
plt.ylabel("Coefficient of variation (%)")
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "cv_dispersion.png", dpi=300, bbox_inches="tight")
plt.close()

# ------------------------------------------------------------
# Figure 2: Mean relevance vs CV
# ------------------------------------------------------------
plt.figure(figsize=(10, 7))

selected = df["Selected"]
not_selected = ~df["Selected"]

plt.scatter(
    df.loc[not_selected, "mu_j"],
    df.loc[not_selected, "CV_j (%)"],
    color="black",
    alpha=0.7,
    label="Not selected"
)

plt.scatter(
    df.loc[selected, "mu_j"],
    df.loc[selected, "CV_j (%)"],
    color="green",
    alpha=0.9,
    label="Selected"
)

for _, row in df.iterrows():
    indicador_num = int(str(row["ID"]).replace("I-", ""))
    plt.text(
        row["mu_j"] + 0.01,
        row["CV_j (%)"] + 0.05,
        str(indicador_num),
        fontsize=8
    )

plt.axvline(
    x=RELEVANCE_THRESHOLD,
    color="blue",
    linestyle="--",
    linewidth=1.5,
    label=f"μ threshold = {RELEVANCE_THRESHOLD}"
)

plt.axhline(
    y=CV_THRESHOLD,
    color="gray",
    linestyle="--",
    linewidth=1.5,
    label=f"CV threshold = {CV_THRESHOLD:.0f}%"
)

plt.xlabel("Mean relevance score (μj)")
plt.ylabel("Coefficient of variation (%)")
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "mean_relevance_vs_cv_labeled.png", dpi=300, bbox_inches="tight")
plt.close()

# ------------------------------------------------------------
# Figure 3: Distribution of mean relevance scores
# ------------------------------------------------------------
plt.figure(figsize=(9, 6))
plt.hist(df["mu_j"], bins=10)
plt.axvline(x=RELEVANCE_THRESHOLD, linestyle="--", label=f"μj = {RELEVANCE_THRESHOLD}")
plt.xlabel("Mean relevance score (μj)")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "mean_relevance_histogram.png", dpi=300, bbox_inches="tight")
plt.close()

# ------------------------------------------------------------
# Figure 4: Initial vs selected indicators by dimension
# ------------------------------------------------------------
x = np.arange(len(summary_by_dimension))
width = 0.35

plt.figure(figsize=(10, 6))
plt.bar(x - width/2, summary_by_dimension["Initial_indicators"], width, label="Initial")
plt.bar(x + width/2, summary_by_dimension["Selected_indicators"], width, label="Selected")
plt.xticks(x, summary_by_dimension["Dimension"], rotation=45, ha="right")
plt.ylabel("Number of indicators")
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "indicators_by_dimension.png", dpi=300, bbox_inches="tight")
plt.close()

# ------------------------------------------------------------
# Figure 5: Sensitivity analysis
# ------------------------------------------------------------
plt.figure(figsize=(8, 5))
plt.plot(
    sensitivity_df["Relevance threshold"],
    sensitivity_df["Selected indicators"],
    marker="o"
)
plt.xlabel("Minimum relevance threshold")
plt.ylabel("Number of selected indicators")
plt.tight_layout()
plt.savefig(FIG_DIR / "sensitivity_relevance_threshold.png", dpi=300, bbox_inches="tight")
plt.close()

print(f"Figures saved in folder: {FIG_DIR}")
```
